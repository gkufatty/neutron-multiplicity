```python
import uproot
import os
```

```python
#this generates the keys of the minirun caf.root file and save them in the folder

# Specify the path to your ROOT file
file_path = '/global/cfs/cdirs/dune/www/data/2x2/simulation/productions/MiniRun6.2_1E19_RHC/MiniRun6.2_1E19_RHC.caf/CAF.flat/0000000/MiniRun6.2_1E19_RHC.caf.0000000.CAF.flat.root'
output_path = '/global/homes/g/gkufatty/projects/neutron-multiplicity/analysis/cfg/mr6.2_keys.txt'
file = uproot.open(file_path)
caftree=file['cafTree']

# Write the keys to the specified file
with open(output_path, 'w') as txt_file:
    for key in caftree.keys():
        txt_file.write(f'{key}\n')

print(f"Keys have been written to {output_path}")
```

```python
directory_path_1= '/global/cfs/cdirs/dune/www/data/2x2/simulation/productions/MiniRun6.2_1E19_RHC/MiniRun6.2_1E19_RHC.caf/CAF.flat/0000000/'
directory_path_2= '/global/cfs/cdirs/dune/www/data/2x2/simulation/productions/MiniRun6.2_1E19_RHC/MiniRun6.2_1E19_RHC.caf/CAF.flat/0001000/'
output_txt_path = '/global/homes/g/gkufatty/projects/neutron-multiplicity/analysis/cfg/minirun6.2.txt'
file_paths = []

for root, dirs, files in os.walk(directory_path_1):
    for file in files:
        file_paths.append(os.path.join(root, file))

for root, dirs, files in os.walk(directory_path_2):
    for file in files:
        file_paths.append(os.path.join(root, file))

file_paths.sort()

# Write all file paths to the output .txt file
with open(output_txt_path, 'w') as txt_file:
    for path in file_paths:
        txt_file.write(f"{path}\n")
```
