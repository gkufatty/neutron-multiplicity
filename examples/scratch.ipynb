1.- numuCC -> Muon track
2.- numu CC in FV (DV) 
3.- target = Ar 
4.- neutron -> neutron-induced proton (We can have more than one) 
( neutron-induced photon) 
5.- Proton conatined in DV 

```python
# Load PDG Codes
pdg_tab = ParticleCode()

plot_dir = './plots-gk/'
# Load datasets 
n_files = 995
location = "nersc"
df=load_dataset(n_files,'nersc')
```

```python
n_vtx=0
count=0
vtx_x = []
vtx_y = []
vtx_z = []

def within_DV(vtx, ev_data):
    vx = ev_data['rec.mc.nu.vtx.x'][vtx]
    vy= ev_data['rec.mc.nu.vtx.y'][vtx]
    vz = ev_data['rec.mc.nu.vtx.z'][vtx] 
    # Geometry for MiniRun 5
    anode_xs = np.array([-63.931, -3.069, 3.069, 63.931])
    anode_ys = np.array([-19.8543, 103.8543]) - 42  # Subtract 42 from all y coordinates
    anode_zs = np.array([-64.3163, -2.6837, 2.6837, 64.3163])

    # Check if the point is within the bounds
    within_x = anode_xs.min() <= vx <= anode_xs.max()
    within_y = anode_ys.min() <= vy <= anode_ys.max()
    within_z = anode_zs.min() <= vz <= anode_zs.max()
    # Return True if the point is within all bounds, otherwise return False
    return within_x and within_y and within_z


for ev in range(len(df)):
    ev_data = df.iloc[ev]  # Use a specific event for this example
    # Grab the PDG code, interaction type, and nucleus type for this event
    ev_pdg = ev_data['rec.mc.nu.pdg']
    ev_type = ev_data['rec.mc.nu.iscc']
    ev_nucleus = ev_data['rec.mc.nu.targetPDG']
    
    # Mask the numus, CC interactions, and Argon targets
    mask_pdg = np.isin(ev_pdg, [pdg_tab.numu, -pdg_tab.numu])  
    mask_cc = (ev_type == 1)  # Get the CC interactions
    mask_Ar = (ev_nucleus == pdg_tab.argon)  # Check for Argon target

    # Find indices where all masks are True
    valid_indices = np.where(mask_pdg & mask_cc & mask_Ar)[0]

    # Only proceed if there are valid indices
    if len(valid_indices) > 0:
        n_vertices = ev_data['rec.mc.nu..length']
        n_prime = ev_data['rec.mc.nu.prim..length'][valid_indices]
        n_seco = ev_data['rec.mc.nu.sec..length'][valid_indices]

        for vtx in valid_indices:
            if within_DV(vtx, ev_data): 
                vtx_x.append(ev_data['rec.mc.nu.vtx.x'][vtx])
                vtx_y.append(ev_data['rec.mc.nu.vtx.y'][vtx])
                vtx_z.append(ev_data['rec.mc.nu.vtx.z'][vtx])
                # # Count number of nu ixn in detector volume 
                n_vtx+=1
                n_prim = ev_data['rec.mc.nu.prim..length'][vtx]
                n_sec = ev_data['rec.mc.nu.sec..length'][vtx]
            else:
                pass

print(f'In {len(df)} events, there are {n_vtx} interactions within the detector ')
```
